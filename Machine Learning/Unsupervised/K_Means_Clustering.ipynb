# K-Means Clustering (Unsupervised Learning)

1. Choose k (predefined by us) random points to be centriod
2. Calculating distance between points and centroid,
assigning point to closest centriod (expectation)
3. Compute new Centroid (maximization)
4. Iterating through step 2 and step 3 to get stable points

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.cluster import KMeans
from sklearn.decomposition import PCA

# Load the data
df = pd.read_csv("seeds_dataset.txt", delim_whitespace=True, header=None)
df.columns = [
    "area", "perimeter", "compactness", "length_of_kernel",
    "width_of_kernel", "asymmetry_coefficient", "length_of_kernel_groove", "class"
]

# K-Means Clustering (using all features)
X = df[df.columns[:-1]].values
kmeans = KMeans(n_clusters=3)
kmeans.fit(X)
clusters = kmeans.labels_
cluster_df = pd.DataFrame(np.hstack((X, clusters.reshape(-1, 1))), columns=df.columns)

print("K-Means Clustering Results:")
display(cluster_df.head())

# Principal Component Analysis
pca = PCA(n_components=2)
transformed_x = pca.fit_transform(X)

print("\nPCA Results:")
print("Original shape:", X.shape)
print("Transformed shape:", transformed_x.shape)
print("Transformed data (first 5 rows):\n", transformed_x[:5])

# K-Means on PCA-transformed data
kmeans_pca = KMeans(n_clusters=3)
kmeans_pca.fit(transformed_x)
kmeans_pca_clusters = kmeans_pca.labels_

kmeans_pca_df = pd.DataFrame(np.hstack((transformed_x, kmeans_pca_clusters.reshape(-1, 1))), columns=["pca1", "pca2", "class"])
truth_pca_df = pd.DataFrame(np.hstack((transformed_x, df["class"].values.reshape(-1, 1))), columns=["pca1", "pca2", "class"])

# Visualize K-Means on PCA
print("\nVisualization of K-Means on PCA-transformed data:")
sns.scatterplot(x="pca1", y="pca2", hue='class', data=kmeans_pca_df)
plt.title("K-Means Clusters on PCA")
plt.show()

# Visualize Original Classes on PCA
print("\nVisualization of Original Classes on PCA-transformed data:")
sns.scatterplot(x="pca1", y="pca2", hue='class', data=truth_pca_df)
plt.title("Original Classes on PCA")
plt.show()
```
