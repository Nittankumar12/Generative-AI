Loss functions in PyTorch
Implement and compare MSE,

Cross-Entropy, and BCE losses with practical techniques




```python
import torch
import torch.nn as nn
import torch.nn.functional as F


# mse- loss for regression
predictions = torch.tensor([2.5, 0.0, 2.0, 8.0])
targets = torch.tensor([3.0, -0.5, 2.0, 7.0])


mse_loss = nn.MSELoss()
loss = mse_loss(predictions, targets)
print(f'mse loss: {loss.item(): 4f}')
# Manual: mean((2.5-3)² + (0-(-0.5))² + (2-2)² + (8-7)²) = 0.375

# huber loss - robust to outliers  (MSE near 0, MAE for large errors)
huber_loss = nn.HuberLoss(delta = 1.0)
print(f'HuberLoss: {huber_loss(predictions, targets):.4f}')

# cross entropy loss - for mult class classification
# raw logits (not softmax probabilities)
logits = torch.tensor([[2.0, 1.0, 0.1],   # Sample 1: class 0 likely
                       [0.5, 2.5, 0.3]])  # Sample 2: class 1 likely
labels = torch.tensor([0, 1])  # True classes (indices, not one-hot)

ce_loss = nn.CrossEntropyLoss()
loss = ce_loss(logits, labels)
print(f"Cross Entropy loss: {loss.item():.4f}")


# what cross entropy does internally
probs = F.softmax(logits, dim = 1)
print(f"Softmax Predictions: \n{probs}")

# label smoothing - prevents over confidence and improves generalization
ce_smooth = nn.CrossEntropyLoss(label_smoothing= 0.1)
loss_smooth = ce_smooth(logits, labels)

print(f"With label smoothing (0.1): {loss_smooth.item():.4f}")

# class weighted ce for imbalanced datasets
# if class 0 has 80% data, class 2 has 5%

weights = torch.tensor([1.0, 5.3, 16.0])
ce_weighted = nn.CrossEntropyLoss(weight = weights)
print(f"class weighted ce - {ce_weighted(logits, labels).item():.4f}")


# ============================================================
# BCE LOSS - For Binary/Multi-label Classification
# ============================================================
# Use BCEWithLogitsLoss for numerical stability (includes sigmoid)

logits = torch.tensor([2.0, -1.0, 0.5])  # Raw scores
targets = torch.tensor([1.0, 0.0, 1.0])  # Binary labels

bce_loss = nn.BCEWithLogitsLoss()
loss = bce_loss(logits, targets)
print(f"BCE loss: {loss.item():.4f}")


weighted_bce = nn.BCEWithLogitsLoss(pos_weight=torch.tensor([99.0]))
loss_weighted = weighted_bce(logits, targets)
print(f"Weighted BCE (99:1): {loss_weighted.item():.4f}")

# ============================================================
# QUICK REFERENCE
# ============================================================
print("\n" + "="*50)
print("LOSS FUNCTION CHEAT SHEET")
print("="*50)
print("Regression:      nn.MSELoss() or nn.L1Loss()")
print("With outliers:   nn.HuberLoss(delta=1.0)")
print("Binary:          nn.BCEWithLogitsLoss() + raw logits")
print("Multi-class:     nn.CrossEntropyLoss() + raw logits")
print("Multi-label:     nn.BCEWithLogitsLoss() + raw logits")
print("Imbalanced:      CrossEntropyLoss(weight=...) or focal loss")
print("Smoothing:       CrossEntropyLoss(label_smoothing=0.1)")
print("="*50)
```
